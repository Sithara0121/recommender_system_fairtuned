```python
# 📦 System Path Setup
import sys
import os
sys.path.append(os.path.abspath('../src'))

# 📊 Imports
import pandas as pd
from content_model import ContentRecommender

# 📁 Load Playlist Data (from previous EDA)
df = pd.read_pickle('../data/processed_playlists.pkl')  # if you saved it earlier from EDA
# If not saved, uncomment below and re-parse it
# df = pd.read_json('../data/spotify_million_playlist.json', lines=True)

# 🧠 Load Track Metadata
df_tracks = pd.read_json('../data/tracks.json', lines=True)

# 🧱 Initialize & Prepare Content-Based Recommender
content_rec = ContentRecommender()
content_rec.prepare(df_tracks)

# 🎧 Choose a playlist
example_pid = df['playlist_id'].sample(1).iloc[0]
playlist_tracks = df[df['playlist_id'] == example_pid]['track_uri'].tolist()

# 🤖 Recommend
print(f"🎯 Content-based recommendations for Playlist ID {example_pid}:")
recommended_tracks = content_rec.recommend_tracks(playlist_tracks, top_n=10)
print("🎵 Recommended Track URIs:")
print(recommended_tracks)
```

```python
import pickle
import os
content_model_path = '../data/models/content_model.pkl'
os.makedirs(os.path.dirname(content_model_path), exist_ok=True)

with open(content_model_path, 'wb') as f:
    pickle.dump(content_rec, f)

print(f"✅ Content-based model saved to: {content_model_path}")
```
